```python
# Step 2: Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Step 3: Load cleaned dataset
df = pd.read_csv("data/cleaned_finance_data.csv")
```

```python
# Step 4: Overview
print("Dataset shape:", df.shape)
print("Columns:", df.columns)
print(df.head())
```

```python
# Step 5: Distribution of key numeric features
numeric_cols = df.select_dtypes(include=np.number).columns
df[numeric_cols].hist(figsize=(14,10), bins=30, color="lightgreen")
plt.suptitle("Distribution of Numeric Features")
plt.show()
```

```python
# Step 6: Liquidity vs Cashflow Volatility
if "liquidity_30d_usd" in df.columns and "cashflow_volatility_pct" in df.columns:
    plt.figure(figsize=(8,6))
    sns.scatterplot(x="liquidity_30d_usd", y="cashflow_volatility_pct", data=df, alpha=0.6)
    plt.title("Liquidity vs Cashflow Volatility")
    plt.show()
```

```python
# Step 8: Persona Archetype vs Net Worth
if "persona_archetype" in df.columns and "net_worth_usd" in df.columns:
    archetype_networth = df.groupby("persona_archetype")["net_worth_usd"].mean().sort_values(ascending=False)
    plt.figure(figsize=(12,6))
    archetype_networth.plot(kind="bar", color="purple")
    plt.title("Average Net Worth by Persona Archetype")
    plt.ylabel("Net Worth (USD)")
    plt.show()
```

```python
# Step 9: Trust vs Engagement
if "trust_score_dynamic_1_100" in df.columns and "engagement_score_1_100" in df.columns:
    plt.figure(figsize=(8,6))
    sns.scatterplot(x="trust_score_dynamic_1_100", y="engagement_score_1_100", data=df, alpha=0.6)
    plt.title("Trust Score vs Engagement Score")
    plt.show()
```

```python
# Step 10: Adherence vs Engagement
if "adherence_score_1_100" in df.columns and "engagement_score_1_100" in df.columns:
    plt.figure(figsize=(8,6))
    sns.scatterplot(x="adherence_score_1_100", y="engagement_score_1_100", data=df, alpha=0.6)
    plt.title("Adherence Score vs Engagement Score")
    plt.show()
```
