```python
# Step 2: Import libraries
import pandas as pd
import numpy as np
```

```python
# Step 3: Load dataset
# Make sure your dataset is saved in the "data" folder as finance_data.csv
df = pd.read_csv("data/personal_finance_customer_data.csv")
```

```python
# Step 4: Inspect data
print("Dataset shape:", df.shape)
print(df.head())
print(df.info())
```

```python
# Step 5: Handle missing values
# Fill missing categories with "Unknown" and missing amounts with 0
df = df.fillna({"Category": "Unknown", "Amount": 0})
```

```python
# Step 5: Handle missing values safely
for col in df.columns:
    if df[col].dtype == "object":   # text columns
        df[col] = df[col].fillna("Unknown")
    else:                           # numeric columns
        df[col] = df[col].fillna(0)
```

```python
# Step 7: Remove duplicates
df = df.drop_duplicates()
```

```python
# Step 8: Save cleaned dataset
df.to_csv("data/cleaned_finance_data.csv", index=False)

print("✅ Data cleaning complete. Cleaned file saved as cleaned_finance_data.csv")
print(df.head())
```
